# **Strategi Pendekatan Masalah Machine Learning (ML)**
Menghadapi masalah ML di dunia nyata memerlukan imajinasi dan ketelitian sistematis. Berdasarkan filosofi Müller & Guido, berikut adalah panduan strategis bagi pengembang:
* Pemahaman Mendalam terhadap Data dan Tugas: Sebelum memilih algoritme, Anda harus memastikan bahwa data yang dikumpulkan benar-benar mengandung informasi yang diperlukan untuk menjawab pertanyaan bisnis atau riset. Melibatkan manusia dalam proses pelabelan berkelanjutan atau peninjauan keputusan kritis memastikan sistem tetap etis dan akurat.

Apakah model ini membantu pengguna mengambil keputusan lebih cepat? Apakah ia mengurangi biaya operasional? Metrik teknis harus selaras dengan dampak dunia nyata.

# **Panduan Membangun Estimator Kustom (Custom Estimator)**
Salah satu kekuatan utama scikit-learn adalah interoperabilitasnya. Dengan membangun estimator kustom, Anda dapat memasukkan logika spesifik domain ke dalam objek yang kompatibel dengan Pipeline, GridSearchCV, dan cross_val_score.


```python
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.utils.validation import check_X_y, check_array, check_is_fitted

class MyCustomClassifier(BaseEstimator, ClassifierMixin):
    """Contoh struktur estimator kustom sesuai standar scikit-learn."""

    def __init__(self, demo_parameter=1):
        # Parameter harus disimpan apa adanya tanpa modifikasi
        # Hal ini penting agar get_params() berfungsi untuk tuning
        self.demo_parameter = demo_parameter

    def fit(self, X, y):
        # Validasi input sesuai standar scikit-learn
        X, y = check_X_y(X, y)

        # Logika pelatihan model diletakkan di sini
        # Atribut yang dipelajari selama pelatihan WAJIB diakhiri dengan '_'
        self.classes_ = np.unique(y)
        self.n_features_in_ = X.shape[1]

        # Menandai bahwa model telah dilatih
        self.is_fitted_ = True

        # Selalu kembalikan self agar bisa dilakukan method chaining
        return self

    def predict(self, X):
        # Pastikan model sudah dilatih sebelum prediksi
        check_is_fitted(self, 'is_fitted_')
        X = check_array(X)

        # Logika prediksi menghasilkan array hasil
        return np.ones(X.shape[0]) # Contoh placeholder$0
```

**Penjelasan Komponen Penting**:
* __init__: Hanya digunakan untuk menyimpan parameter. Jangan melakukan pemrosesan data di sini agar kompatibel dengan fungsi get_params() yang digunakan dalam Grid Search.
* fit: Tempat logika pembelajaran utama. Penggunaan trailing underscore (seperti self.classes_) adalah konvensi penting dalam scikit-learn untuk membedakan antara parameter input pengguna dan atribut yang dipelajari dari data.
* predict: Menghasilkan keluaran berdasarkan data input. Memastikan model telah melalui proses fit menggunakan check_is_fitted.
