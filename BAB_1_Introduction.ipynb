#**NumPy**

* **Fungsi Utama**: Pustaka dasar di Python yang digunakan untuk kebutuhan komputasi numerik, pengelolaan array multidimensi (ndarray), dan operasi matematika tingkat tinggi.

* **Peran dalam Scikit-Learn**: Array NumPy merupakan struktur data standar dan utama di scikit-learn. Semua data mentah wajib diubah ke format ini sebelum diproses oleh model.

* **Karakteristik Data**: Struktur ndarray bersifat homogen, yang berarti seluruh elemen di dalam array tersebut harus memiliki tipe data yang sama (seragam).




```python
import numpy as np
x = np.array([[1, 2, 3], [4, 5, 6]])
print("x:\n{}".format(x))
```

# **SciPy**
* **Peran SciPy**: Kumpulan fungsi komputasi ilmiah lanjutan yang menyediakan algoritma dasar (seperti aljabar linier dan statistik) yang diadopsi oleh scikit-learn.

* **Fungsi Matriks Jarang (Sparse Matrix)**: Digunakan untuk menyimpan data dua dimensi (2D) yang mayoritas elemennya bernilai nol ($0$).

* **Efisiensi Memori**: Dibandingkan menyimpan semua angka nol yang memenuhi memori (dense representation), sparse matrix hanya mencatat koordinat posisi dan nilai dari elemen yang bukan nol saja.

```python
from scipy import sparse
# Create a 2D NumPy array with a diagonal of ones, and zeros everywhere else
eye = np.eye(4)
print("NumPy array:\n{}".format(eye))
```

```python
# Convert the NumPy array to a SciPy sparse matrix in CSR format
# Only the nonzero entries are stored
sparse_matrix = sparse.csr_matrix(eye)
print("\nSciPy sparse CSR matrix:\n{}".format(sparse_matrix))
```

```python
data = np.ones(4)
row_indices = np.arange(4)
col_indices = np.arange(4)
eye_coo = sparse.coo_matrix((data, (row_indices, col_indices)))
print("COO representation:\n{}".format(eye_coo))
```

# **Training and Testing Data**
Untuk mengevaluasi apakah model dapat melakukan generalisasi pada data baru, kita membagi dataset menjadi dua bagian. Menggunakan proporsi 75% untuk pelatihan dan 25% untuk pengujian dianggap sebagai "rule of thumb" (aturan praktis) yang baik untuk dataset berukuran kecil.
Dalam penamaan variabel, kita menggunakan X kapital untuk matriks input karena merupakan array dua dimensi, dan y kecil untuk vektor target karena merupakan array satu dimensi. Konvensi ini mengikuti notasi matematika f(x)=y.

```python
from sklearn.model_selection import train_test_split

# Membagi data dengan parameter random_state=0 untuk hasil yang konsisten
X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'], iris_dataset['target'], random_state=0)

# Komentar: Secara default, fungsi ini mengalokasikan 75% data sebagai training set
# dan 25% sisanya sebagai test set.
```

# **k-Nearest Neighbors**
Algoritma k-NN membangun model dengan cara menyimpan data pelatihan. Prediksi dibuat dengan mencari titik terdekat (tetangga) pada ruang fitur. Di sini, kita menggunakan n_neighbors=1 untuk kesederhanaan.


```python
from sklearn.neighbors import KNeighborsClassifier

# Instansiasi model k-NN dengan satu tetangga terdekat
knn = KNeighborsClassifier(n_neighbors=1)

# Membangun model menggunakan algoritma k-NN pada data pelatihan
knn.fit(X_train, y_train)
```

# **Proses Prediksi Data Baru**
Setelah model terlatih, kita dapat melakukan prediksi pada sampel baru. Perlu diingat bahwa scikit-learn selalu mengharuskan data input dalam bentuk array dua dimensi (matriks), bahkan jika hanya ada satu sampel. Itulah sebabnya kita menggunakan tanda kurung siku ganda [[]].

```python
# Data baru: sepal length 5cm, sepal width 2.9cm, petal length 1cm, petal width 0.2cm
X_new = np.array([[5, 2.9, 1, 0.2]])

# Melakukan prediksi berdasarkan data baru tersebut
prediction = knn.predict(X_new)

print("Hasil Prediksi (integer): {}".format(prediction))
print("Nama spesies hasil prediksi: {}".format(
    iris_dataset['target_names'][prediction]))
```

# **Evaluasi Akurasi Model**
Akurasi merupakan metrik untuk mengukur fraksi atau proporsi prediksi yang benar dibandingkan dengan total jumlah sampel pada set pengujian. Jika model menghasilkan skor 0.97, ini berarti kita bisa mengharapkan model benar sebanyak 97% dari waktu saat menemui bunga Iris baru di masa depan.

```python
# Metode 1: Menghitung rata-rata kecocokan antara prediksi dan label asli secara manual
y_pred = knn.predict(X_test)
print("Akurasi test set (manual): {:.2f}".format(np.mean(y_pred == y_test)))

# Metode 2: Menggunakan metode .score langsung pada set pengujian
print("Akurasi test set (score method): {:.2f}".format(knn.score(X_test, y_test)))
```
